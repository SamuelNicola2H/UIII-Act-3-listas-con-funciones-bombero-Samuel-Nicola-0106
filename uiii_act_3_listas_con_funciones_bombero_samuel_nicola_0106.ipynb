### Ejemplo funciones lista Bomberos

![link text](https://raw.githubusercontent.com/SamuelNicola2H/UIII-Act-3-listas-con-funciones-bombero-Samuel-Nicola-0106/refs/heads/main/infografia%20bomberos.png)

```python
# Inicialización de la lista de bomberos (diccionario vacío al inicio)
estacion_bomberos = []

def agregar_bombero():
    print("\n--- Registrar Bombero ---")
    bombero = {}
    bombero["nombre"] = input("Nombre: ")
    bombero["obras"] = input("Obras/Servicios realizados: ")
    bombero["premios"] = input("Premios o reconocimientos: ")
    bombero["habilidades"] = input("Habilidades especiales: ")
    bombero["edad"] = input("Rango de edad: ")
    bombero["ultimo"] = input("Último proyecto/incidente: ")
    estacion_bomberos.append(bombero)
    print("¡Bombero registrado con éxito!")
    listar_bomberos()

def listar_bomberos():
    if not estacion_bomberos:
        print("\n[!] No hay bomberos registrados.")
        return

    print("\n" + "="*125)
    print(f"{'NOMBRE':<15} | {'OBRAS':<15} | {'PREMIOS':<15} | {'HABILIDADES':<20} | {'EDAD':<6} | {'ÚLTIMO PROYECTO'}")
    print("-" * 125)
    for b in estacion_bomberos:
        print(f"{b['nombre']:<15} | {b['obras']:<15} | {b['premios']:<15} | {b['habilidades']:<20} | {b['edad']:<6} | {b['ultimo']}")
    print("="*125)

def buscar_bombero():
    nombre = input("\nIngrese el nombre a buscar: ")
    for b in estacion_bomberos:
        if b["nombre"].lower() == nombre.lower():
            print("\n--- Información Detallada ---")
            for clave, valor in b.items():
                print(f"{clave.upper()}: {valor}")
            return
    print("Bombero no encontrado.")

def borrar_bombero():
    nombre = input("\nNombre del bombero a dar de baja: ")
    for i in range(len(estacion_bomberos)):
        if estacion_bomberos[i]["nombre"].lower() == nombre.lower():
            estacion_bomberos.pop(i)
            print("Bombero eliminado del sistema.")
            listar_bomberos()
            return
    print("Bombero no encontrado.")

def actualizar_bombero():
    nombre = input("\nNombre del bombero a actualizar: ")
    for b in estacion_bomberos:
        if b["nombre"].lower() == nombre.lower():
            print("Ingrese los nuevos datos (deje vacío para mantener):")
            b["obras"] = input(f"Obras ({b['obras']}): ") or b["obras"]
            b["ultimo"] = input(f"Último proyecto ({b['ultimo']}): ") or b["ultimo"]
            print("Datos actualizados.")
            listar_bomberos()
            return
    print("Bombero no encontrado.")

# Menú Principal
while True:
    print("\n--- SISTEMA DE GESTIÓN DE BOMBEROS ---")
    print("1. Agregar bombero\n2. Listar bomberos\n3. Buscar bombero\n4. Borrar bombero\n5. Actualizar bombero\n6. Salir")
    opcion = input("Seleccione una opción: ")

    if opcion == "1": agregar_bombero()
    elif opcion == "2": listar_bomberos()
    elif opcion == "3": buscar_bombero()
    elif opcion == "4": borrar_bombero()
    elif opcion == "5": actualizar_bombero()
    elif opcion == "6":
        print("Cerrando sistema... Samuel nicola = 40")
        break
    else: print("Opción no válida.")
```
