### Ejemplo funciones lista Bomberos

```python
# 1. Inicialización: Lista de listas para almacenar los datos
# Estructura interna: [[id1, nombre1], [id2, nombre2]]
bomberos = []

# 2. Función para añadir un nuevo registro
def agregar():
    id_b = input("Ingrese el ID del bombero: ")
    nombre = input("Ingrese el nombre: ")
    # Añadimos una sublista con los dos datos
    bomberos.append([id_b, nombre])
    print(f"Bombero '{nombre}' agregado con éxito.")

# 3. Función para mostrar todos los registros
def listar():
    print("\n" + "="*30)
    print("      LISTA DE BOMBEROS")
    print("="*30)
    if not bomberos:
        print("La lista está vacía.")
    else:
        # Recorremos la lista principal y desempaquetamos la sublista
        for b in bomberos:
            print(f"ID: {b[0]} | Nombre: {b[1]}")
    print("="*30)

# 4. Función de apoyo para localizar el índice de un bombero
def buscar_indice():
    objetivo = input("Ingrese el ID a buscar: ")
    for i in range(len(bomberos)):
        if bomberos[i][0] == objetivo: # Comparamos con la posición 0 (el ID)
            return i # Retornamos la posición en la lista
    return -1

# 5. Función para visualizar detalle específico
def leer():
    indice = buscar_indice()
    if indice != -1:
        b = bomberos[indice]
        print(f"\nDETALLE ENCONTRADO:\nID: {b[0]}\nNombre: {b[1]}")
    else:
        print("Error: El ID no existe en el sistema.")

# 6. Función para modificar (Update)
def modificar():
    indice = buscar_indice()
    if indice != -1:
        nuevo_nombre = input("Ingrese el nuevo nombre para este ID: ")
        bomberos[indice][1] = nuevo_nombre # Modificamos el elemento en la posición 1
        print("Registro actualizado correctamente.")
    else:
        print("No se puede modificar: ID no encontrado.")

# 7. Función para eliminar (Delete)
def borrar():
    indice = buscar_indice()
    if indice != -1:
        # Usamos pop con el índice encontrado
        eliminado = bomberos.pop(indice)
        print(f"Registro eliminado: {eliminado[1]} (ID: {eliminado[0]})")
    else:
        print("No se encontró el registro para eliminar.")

# --- MENÚ PRINCIPAL ---
def menu():
    while True:
        print("\n--- SISTEMA DE GESTIÓN DE BOMBEROS ---")
        print("1. Agregar")
        print("2. Listar todos")
        print("3. Buscar (ver nombre)")
        print("4. Leer (detalle completo)")
        print("5. Modificar")
        print("6. Borrar")
        print("7. Salir")

        opcion = input("\nSeleccione una opción (1-7): ")

        if opcion == "1":
            agregar()
        elif opcion == "2":
            listar()
        elif opcion == "3":
            idx = buscar_indice()
            if idx != -1: print(f"Resultado: {bomberos[idx][1]}")
            else: print("No encontrado.")
        elif opcion == "4":
            leer()
        elif opcion == "5":
            modificar()
        elif opcion == "6":
            borrar()
        elif opcion == "7":
            print("Finalizando programa. ¡Buen turno!")
            break
        else:
            print("Opción no válida, intente de nuevo.")

# Ejecución del programa
if __name__ == "__main__":
    menu()
```
